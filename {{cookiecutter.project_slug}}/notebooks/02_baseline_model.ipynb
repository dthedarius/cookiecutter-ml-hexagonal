```python
import sys
from pathlib import Path

# Add project root to path
project_root = Path("..").resolve()
sys.path.insert(0, str(project_root / "src"))
sys.path.insert(0, str(project_root))
```

```python
import json

from {{ cookiecutter.project_slug }}.adapters.outbound.baseline_model import BaselineModel
from pipelines.evaluation.evaluate import compute_metrics

# Load test data
test_path = project_root / "data" / "processed" / "test.json"
if not test_path.exists():
    print("Run 'make generate-data && make preprocess' first.")
else:
    with test_path.open() as f:
        test_data = json.load(f)
    print(f"Loaded {len(test_data)} test samples")
```

## 1. Baseline Model Evaluation

```python
# Initialize and evaluate baseline model
model = BaselineModel()
eval_metrics = model.evaluate(test_data)

print("Baseline Metrics:")
for k, v in eval_metrics.items():
    print(f"  {k}: {v:.4f}")
```

## 2. Per-Sample Analysis

```python
# Analyze predictions sample by sample
from {{ cookiecutter.project_slug }}.domain.value_objects.prediction_label import PredictionLabel

labels = list(PredictionLabel)
print(f"Available labels: {[l.value for l in labels]}")

# Show a few predictions
for sample in test_data[:5]:
    raw_scores = model.predict(sample.get("text", str(sample)))
    best_label = max(raw_scores, key=raw_scores.get)
    print(f"  Input: {str(sample)[:60]}...  Prediction: {best_label}  Scores: {raw_scores}")
```
