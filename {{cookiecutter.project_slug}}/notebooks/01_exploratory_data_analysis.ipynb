```python
import json
from pathlib import Path

import pandas as pd

# Load raw data
DATA_DIR = Path("..") / "data" / "raw"
data_files = list(DATA_DIR.glob("*.json"))
print(f"Found {len(data_files)} data file(s): {[f.name for f in data_files]}")
```

```python
# Load data into a DataFrame
all_data = []
for f in data_files:
    with f.open() as fh:
        all_data.extend(json.load(fh))

df = pd.DataFrame(all_data)
print(f"Dataset shape: {df.shape}")
df.head()
```

## 1. Data Overview

```python
# Data types and info
df.info()
```

```python
# Missing values
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
print("Missing values:")
print(pd.DataFrame({"count": missing, "percent": missing_pct}))
```

```python
# Duplicates
n_duplicates = df.duplicated().sum()
print(f"Duplicate rows: {n_duplicates} ({n_duplicates / len(df) * 100:.1f}%)")
```

```python
# Descriptive statistics
df.describe(include="all")
```

## 2. Target Distribution

```python
{% if cookiecutter.ml_task == 'regression' %}
# Regression target distribution
print(df["target"].describe())
df["target"].hist(bins=30)
{% else %}
# Classification label distribution
label_col = "label"
print(df[label_col].value_counts())
print(f"\nClass balance ratio: {df[label_col].value_counts(normalize=True).to_dict()}")
df[label_col].value_counts().plot(kind="bar", title="Label Distribution")
{% endif %}
```

## 3. Feature Analysis

```python
# Text length distribution
if "text" in df.columns:
    df["text_length"] = df["text"].str.len()
    print(df["text_length"].describe())
    df["text_length"].hist(bins=30, alpha=0.7)
    print(f"\nEmpty texts: {(df['text_length'] == 0).sum()}")
```
